# 1️⃣ Optimización de Asignación de Ejecutivos a Clientes

## Contexto
Optimizar la asignación cliente–ejecutivo maximizando rentabilidad y minimizando penalizaciones por distancia y sobrecarga.

## Función Objetivo
Maximizar:
Z = Σ (R_i * P_ij) − αD_ij − βS_j

Sujeto a restricciones de capacidad y segmentación.

## Estructura del Cromosoma
- Gen: Ejecutivo asignado a un cliente
- Genotipo: Vector numérico
- Fenotipo: Asignación real cliente–ejecutivo

```python
import numpy as np

num_clientes = 10
num_ejecutivos = 3

rentabilidad = np.random.uniform(1000, 5000, num_clientes)
probabilidad = np.random.uniform(0.5, 1.0, (num_clientes, num_ejecutivos))
distancia = np.random.uniform(1, 10, (num_clientes, num_ejecutivos))

alpha = 10
beta = 50
capacidad_max = 5

cromosoma = np.random.randint(0, num_ejecutivos, num_clientes)

def funcion_objetivo(cromosoma):
    total = 0
    penalizacion = 0
    
    for i in range(num_clientes):
        j = cromosoma[i]
        total += rentabilidad[i] * probabilidad[i][j]
        total -= alpha * distancia[i][j]
    
    for j in range(num_ejecutivos):
        carga = np.sum(cromosoma == j)
        if carga > capacidad_max:
            penalizacion += beta * (carga - capacidad_max)
    
    return total - penalizacion

print("Cromosoma:", cromosoma)
print("Fitness:", funcion_objetivo(cromosoma))
```

# 2️⃣ Optimización de Horarios de Cajeros

## Función Objetivo
Minimizar:
Z = C_personal + C_espera + C_horas_extra

## Estructura del Cromosoma
- Gen: Asignación de cajero en turno
- Genotipo: Matriz binaria
- Fenotipo: Calendario real de turnos

```python
num_cajeros = 5
num_turnos = 8

costo_turno = 100
costo_espera = 50
demanda_minima = np.random.randint(1, 4, num_turnos)

cromosoma = np.random.randint(0, 2, (num_cajeros, num_turnos))

def fitness_horarios(cromosoma):
    costo_total = 0
    
    for t in range(num_turnos):
        asignados = np.sum(cromosoma[:, t])
        costo_total += asignados * costo_turno
        
        if asignados < demanda_minima[t]:
            costo_total += costo_espera * (demanda_minima[t] - asignados)
    
    return costo_total

print("Fitness horario:", fitness_horarios(cromosoma))
```

# 3️⃣ Optimización de Parámetros de Scoring Crediticio

## Función Objetivo
Maximizar:
Z = Ingresos − (PD × LGD × EAD)

## Estructura del Cromosoma
- Gen: Peso de variable
- Genotipo: Vector real
- Fenotipo: Modelo de scoring aplicado

```python
num_variables = 5
cromosoma = np.random.dirichlet(np.ones(num_variables))

ingresos = np.random.uniform(10000, 50000)
pd = np.random.uniform(0.01, 0.1)
lgd = 0.4
ead = 20000

def fitness_scoring(cromosoma):
    perdida_esperada = pd * lgd * ead
    utilidad = ingresos - perdida_esperada
    return utilidad

print("Pesos del modelo:", cromosoma)
print("Fitness scoring:", fitness_scoring(cromosoma))
```
